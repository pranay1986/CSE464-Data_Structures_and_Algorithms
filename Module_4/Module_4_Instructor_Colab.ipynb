## 01. Binary Tree Traversals

**Source classification:** Traversals

**Algorithm:** Visit root recursively in pre/in/post order.

**Implementation logic:** Use standard node objects.

**Example input / execution:** `t=Node(2,Node(1),Node(3));print(preorder(t),inorder(t),postorder(t))`

**Expected output:** `[2, 1, 3] [1, 2, 3] [1, 3, 2]`

**Dry-run trace:** pre root-left-right; in left-root-right; post left-right-root

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Binary Tree Traversals.
Algorithm: Visit root recursively in pre/in/post order.
Logic: Use standard node objects.
"""

class Node:
    def __init__(self,x,left=None,right=None):self.x=x;self.left=left;self.right=right
def preorder(t):return [] if not t else [t.x]+preorder(t.left)+preorder(t.right)
def inorder(t):return [] if not t else inorder(t.left)+[t.x]+inorder(t.right)
def postorder(t):return [] if not t else postorder(t.left)+postorder(t.right)+[t.x]


t=Node(2,Node(1),Node(3));print(preorder(t),inorder(t),postorder(t))
```

## 02. Binary Tree Level Order

**Source classification:** Tree traversal

**Algorithm:** Use FIFO queue for breadth-first tree traversal.

**Implementation logic:** Append children in left-to-right order.

**Example input / execution:** `print(levels(Node(1,Node(2),Node(3))))`

**Expected output:** `[1, 2, 3]`

**Dry-run trace:** queue [1] -> [2,3] -> [3] -> []

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Binary Tree Level Order.
Algorithm: Use FIFO queue for breadth-first tree traversal.
Logic: Append children in left-to-right order.
"""

from collections import deque
class Node:
    def __init__(self,x,left=None,right=None):self.x=x;self.left=left;self.right=right
def levels(root):
    if not root:return []
    q=deque([root]);result=[]
    while q:
        node=q.popleft();result.append(node.x)
        if node.left:q.append(node.left)
        if node.right:q.append(node.right)
    return result


print(levels(Node(1,Node(2),Node(3))))
```

## 03. Bst Operations

**Source classification:** BST operations

**Algorithm:** Maintain left smaller, right larger; delete using inorder successor.

**Implementation logic:** Duplicates ignored.

**Example input / execution:** `t=None; for x in (5,3,7,2,4,6,8):t=insert(t,x); t=delete(t,5);print(inorder(t),search(t,6))`

**Expected output:** `[2, 3, 4, 6, 7, 8] True`

**Dry-run trace:** delete 5 (two children): replace by successor 6, remove old 6

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Bst Operations.
Algorithm: Maintain left smaller, right larger; delete using inorder successor.
Logic: Duplicates ignored.
"""

class Node:
    def __init__(self,x):self.x=x;self.left=self.right=None
def insert(root,x):
    if not root:return Node(x)
    if x<root.x:root.left=insert(root.left,x)
    elif x>root.x:root.right=insert(root.right,x)
    return root
def search(root,x):
    while root:
        if x==root.x:return True
        root=root.left if x<root.x else root.right
    return False
def delete(root,x):
    if not root:return None
    if x<root.x:root.left=delete(root.left,x)
    elif x>root.x:root.right=delete(root.right,x)
    else:
        if not root.left:return root.right
        if not root.right:return root.left
        p=root.right
        while p.left:p=p.left
        root.x=p.x;root.right=delete(root.right,p.x)
    return root
def inorder(root):return inorder(root.left)+[root.x]+inorder(root.right) if root else []


t=None
for x in (5,3,7,2,4,6,8):t=insert(t,x)
t=delete(t,5);print(inorder(t),search(t,6))
```

**Example input / execution:** `t=Node(1,Node(2,Node(4)),Node(3));print(size(t),height(t))`

**Expected output:** `4 2`

**Dry-run trace:** leaf height0; node2 height1; root height2

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Tree Height Size.
Algorithm: Recursively count nodes and maximal edge-depth.
Logic: Empty height -1, singleton height 0.
"""

class Node:
    def __init__(self,x,l=None,r=None):self.x=x;self.left=l;self.right=r
def size(t):return 0 if t is None else 1+size(t.left)+size(t.right)
def height(t):return -1 if t is None else 1+max(height(t.left),height(t.right))


t=Node(1,Node(2,Node(4)),Node(3));print(size(t),height(t))
```

## 05. Reconstruct Tree

**Source classification:** Reconstruction exercise

**Algorithm:** Build tree from preorder and inorder of distinct values.

**Implementation logic:** Track inorder indices and consume preorder.

**Example input / execution:** `t=build([2,1,3],[1,2,3]);print(postorder(t))`

**Expected output:** `[1, 3, 2]`

**Dry-run trace:** pre root2; in splits [1] and [3]

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Reconstruct Tree.
Algorithm: Build tree from preorder and inorder of distinct values.
Logic: Track inorder indices and consume preorder.
"""

class Node:
    def __init__(self,x):self.x=x;self.left=self.right=None
def build(preorder,inorder):
    if len(preorder)!=len(inorder) or len(set(inorder))!=len(inorder) or set(preorder)!=set(inorder):raise ValueError('distinct matching keys required')
    index={v:i for i,v in enumerate(inorder)};it=iter(preorder)
    def rec(lo,hi):
        if lo>hi:return None
        v=next(it);mid=index[v]
        if not lo<=mid<=hi:raise ValueError('inconsistent traversals')
        n=Node(v);n.left=rec(lo,mid-1);n.right=rec(mid+1,hi);return n
    return rec(0,len(inorder)-1)
def postorder(t):return postorder(t.left)+postorder(t.right)+[t.x] if t else []


t=build([2,1,3],[1,2,3]);print(postorder(t))
```

## 06. Expression Tree

**Source classification:** Expression tree

**Algorithm:** Construct tree from space-separated postfix tokens.

**Implementation logic:** Operators pop two subtrees; evaluate recursively.

**Example input / execution:** `print(evaluate(build('2 3 4 * +')))`

**Expected output:** `14.0`

**Dry-run trace:** push 2,3,4; combine 3*4; combine 2+12

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Expression Tree.
Algorithm: Construct tree from space-separated postfix tokens.
Logic: Operators pop two subtrees; evaluate recursively.
"""

class Node:
    def __init__(self,x,l=None,r=None):self.x=x;self.left=l;self.right=r
def build(postfix):
    s=[]
    for x in postfix.split():
        if x in ('+','-','*','/'):
            r=s.pop();l=s.pop();s.append(Node(x,l,r))
        else:s.append(Node(float(x)))
    if len(s)!=1:raise ValueError('malformed expression')
    return s[0]
def evaluate(n):
    if n.left is None:return n.x
    a=evaluate(n.left);b=evaluate(n.right)
    return {'+':lambda:a+b,'-':lambda:a-b,'*':lambda:a*b,'/':lambda:a/b}[n.x]()


print(evaluate(build('2 3 4 * +')))
```

## 07. Avl Tree

**Source classification:** AVL insert/rotations

**Algorithm:** BST insert with height maintenance and LL/RR/LR/RL rotations.

**Implementation logic:** Rebalance whenever absolute balance factor exceeds one.

**Example input / execution:** `r=None; for v in (30,20,10,25,40,35):r=insert(r,v); print(r.x,inorder(r))`

**Expected output:** `30 [10, 20, 25, 30, 35, 40]`

**Dry-run trace:** 30,20,10 cause LL rotation; remaining insertions rebalance

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Avl Tree.
Algorithm: BST insert with height maintenance and LL/RR/LR/RL rotations.
Logic: Rebalance whenever absolute balance factor exceeds one.
"""

class Node:
    def __init__(self,x):self.x=x;self.left=self.right=None;self.h=1
def h(n):return n.h if n else 0
def update(n):n.h=1+max(h(n.left),h(n.right))
def right(y):
    x=y.left;y.left=x.right;x.right=y;update(y);update(x);return x
def left(x):
    y=x.right;x.right=y.left;y.left=x;update(x);update(y);return y
def insert(n,x):
    if not n:return Node(x)
    if x<n.x:n.left=insert(n.left,x)
    elif x>n.x:n.right=insert(n.right,x)
    else:return n
    update(n);b=h(n.left)-h(n.right)
    if b>1:
        if x>n.left.x:n.left=left(n.left)
        return right(n)
    if b<-1:
        if x<n.right.x:n.right=right(n.right)
        return left(n)
    return n
def inorder(n):return inorder(n.left)+[n.x]+inorder(n.right) if n else []


r=None
for v in (30,20,10,25,40,35):r=insert(r,v)
print(r.x,inorder(r))
```

## 08. Right Threaded Tree

**Source classification:** Threaded binary tree

**Algorithm:** Thread null right links to inorder successor.

**Implementation logic:** Build a static threaded representation from BST and traverse without stack.

**Example input / execution:** `r=None; for v in (4,2,6,1,3,5,7):r=insert(r,v); thread_tree(r);print(traverse(r))`

**Expected output:** `[1, 2, 3, 4, 5, 6, 7]`

**Dry-run trace:** inorder successor threads replace null right links

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Right Threaded Tree.
Algorithm: Thread null right links to inorder successor.
Logic: Build a static threaded representation from BST and traverse without stack.
"""

class Node:
    def __init__(self,x):self.x=x;self.left=self.right=None;self.thread=False
def insert(root,x):
    if not root:return Node(x)
    p=root
    while True:
        if x<p.x:
            if p.left:p=p.left
            else:p.left=Node(x);return root
        else:
            if p.right:p=p.right
            else:p.right=Node(x);return root
def thread_tree(root):
    prev=[None]
    def walk(n):
        if not n:return
        walk(n.left)
        if prev[0] and prev[0].right is None:prev[0].right=n;prev[0].thread=True
        prev[0]=n;walk(n.right)
    walk(root)
def traverse(root):
    def leftmost(n):
        while n and n.left:n=n.left
        return n
    p=leftmost(root);out=[]
    while p:
        out.append(p.x)
        p=p.right if p.thread else leftmost(p.right)
    return out


r=None
for v in (4,2,6,1,3,5,7):r=insert(r,v)
thread_tree(r);print(traverse(r))
```

## 09. B Tree

**Source classification:** B-tree core insertion/search

**Algorithm:** Implement minimum-degree B-tree insertion with splitting and search.

**Implementation logic:** Minimum degree t means max 2t-1 keys; duplicates ignored.

**Example input / execution:** `b=BTree(2); for x in (10,20,5,6,12,30,7,17):b.insert(x); print(b.traverse(),b.search(12))`

**Expected output:** `[5, 6, 7, 10, 12, 17, 20, 30] True`

**Dry-run trace:** split full root during insertion; inorder keys sorted

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | B Tree.
Algorithm: Implement minimum-degree B-tree insertion with splitting and search.
Logic: Minimum degree t means max 2t-1 keys; duplicates ignored.
"""

class Node:
    def __init__(self,leaf=True):self.keys=[];self.child=[];self.leaf=leaf
class BTree:
    def __init__(self,t=2):
        if t<2:raise ValueError('t>=2')
        self.t=t;self.root=Node()
    def search(self,k,node=None):
        n=node or self.root;i=0
        while i<len(n.keys) and k>n.keys[i]:i+=1
        if i<len(n.keys) and n.keys[i]==k:return True
        return False if n.leaf else self.search(k,n.child[i])
    def split(self,p,i):
        t=self.t;y=p.child[i];z=Node(y.leaf);median=y.keys[t-1]
        z.keys=y.keys[t:];y.keys=y.keys[:t-1]
        if not y.leaf:z.child=y.child[t:];y.child=y.child[:t]
        p.keys.insert(i,median);p.child.insert(i+1,z)
    def nonfull(self,n,k):
        i=len(n.keys)-1
        if n.leaf:
            n.keys.append(k)
            while i>=0 and k<n.keys[i]:n.keys[i+1]=n.keys[i];i-=1
            n.keys[i+1]=k
        else:
            while i>=0 and k<n.keys[i]:i-=1
            i+=1
            if len(n.child[i].keys)==2*self.t-1:
                self.split(n,i)
                if k>n.keys[i]:i+=1
            self.nonfull(n.child[i],k)
    def insert(self,k):
        if self.search(k):return
        r=self.root
        if len(r.keys)==2*self.t-1:
            s=Node(False);s.child=[r];self.root=s;self.split(s,0);self.nonfull(s,k)
        else:self.nonfull(r,k)
    def traverse(self,n=None):
        n=n or self.root;out=[]
        for i,k in enumerate(n.keys):
            if not n.leaf:out+=self.traverse(n.child[i])
            out.append(k)
        if not n.leaf:out+=self.traverse(n.child[-1])
        return out


b=BTree(2)
for x in (10,20,5,6,12,30,7,17):b.insert(x)
print(b.traverse(),b.search(12))
```

## 10. Bplus Tree

**Source classification:** B+ insertion / range challenge

**Algorithm:** Insert keys into linked leaves and split up to the root.

**Implementation logic:** Max_keys keys/leaf, internal separators equal first keys of right subtrees; no duplicates.

**Example input / execution:** `t=BPlusTree(); for x in (10,20,5,6,12,30,7,17):t.insert(x); print(t.range_search(6,20),t.search(12))`

**Expected output:** `[6, 7, 10, 12, 17, 20] True`

**Dry-run trace:** search leaf containing 6, scan linked leaves through 20

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Bplus Tree.
Algorithm: Insert keys into linked leaves and split up to the root.
Logic: Max_keys keys/leaf, internal separators equal first keys of right subtrees; no duplicates.
"""

class Node:
    def __init__(self,leaf=False):self.leaf=leaf;self.keys=[];self.children=[];self.next=None
class BPlusTree:
    def __init__(self,max_keys=3):
        if max_keys<2:raise ValueError('max_keys>=2')
        self.max_keys=max_keys;self.root=Node(True)
    def _leaf(self,key):
        n=self.root
        while not n.leaf:
            i=0
            while i<len(n.keys) and key>=n.keys[i]:i+=1
            n=n.children[i]
        return n
    def search(self,key):return key in self._leaf(key).keys
    def _insert(self,n,key):
        if n.leaf:
            i=0
            while i<len(n.keys) and n.keys[i]<key:i+=1
            n.keys.insert(i,key)
            if len(n.keys)<=self.max_keys:return None
            mid=(len(n.keys)+1)//2;right=Node(True);right.keys=n.keys[mid:];n.keys=n.keys[:mid];right.next=n.next;n.next=right
            return right.keys[0],right
        i=0
        while i<len(n.keys) and key>=n.keys[i]:i+=1
        split=self._insert(n.children[i],key)
        if split is None:return None
        sep,right=split;n.keys.insert(i,sep);n.children.insert(i+1,right)
        if len(n.keys)<=self.max_keys:return None
        mid=len(n.keys)//2;promote=n.keys[mid];new=Node();new.keys=n.keys[mid+1:];new.children=n.children[mid+1:];n.keys=n.keys[:mid];n.children=n.children[:mid+1]
        return promote,new
    def insert(self,key):
        if self.search(key):return
        split=self._insert(self.root,key)
        if split:
            sep,right=split;r=Node();r.keys=[sep];r.children=[self.root,right];self.root=r
    def range_search(self,lo,hi):
        if lo>hi:return []
        n=self._leaf(lo);res=[]
        while n:
            for x in n.keys:
                if x>hi:return res
                if x>=lo:res.append(x)
            n=n.next
        return res


t=BPlusTree()
for x in (10,20,5,6,12,30,7,17):t.insert(x)
print(t.range_search(6,20),t.search(12))
```
