## 01. Singly Linked List

**Source classification:** Core operations

**Algorithm:** Maintain linked nodes with head reference.

**Implementation logic:** Implement insertion, removal and traversal.

**Example input / execution:** `a=SinglyList();[a.append(x) for x in (10,20,30)];a.prepend(5);a.delete(20);print(a.values(),a.search(30))`

**Expected output:** `[5, 10, 30] 2`

**Dry-run trace:** 10 -> 20 -> 30; prepend 5; bypass 20

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Singly Linked List.
Algorithm: Maintain linked nodes with head reference.
Logic: Implement insertion, removal and traversal.
"""

class Node:
    def __init__(self,x,next=None):self.data=x;self.next=next
class SinglyList:
    def __init__(self):self.head=None
    def append(self,x):
        if not self.head:self.head=Node(x);return
        p=self.head
        while p.next:p=p.next
        p.next=Node(x)
    def prepend(self,x):self.head=Node(x,self.head)
    def search(self,x):
        p=self.head;i=0
        while p:
            if p.data==x:return i
            p=p.next;i+=1
        return -1
    def delete(self,x):
        dummy=Node(None,self.head);p=dummy
        while p.next and p.next.data!=x:p=p.next
        if p.next:p.next=p.next.next
        self.head=dummy.next
    def values(self):
        result=[];p=self.head
        while p:result.append(p.data);p=p.next
        return result


a=SinglyList();[a.append(x) for x in (10,20,30)];a.prepend(5);a.delete(20);print(a.values(),a.search(30))
```

## 02. Linked Stack

**Source classification:** Linked stack

**Algorithm:** Use list head as stack top.

**Implementation logic:** Push/pop in O(1).

**Example input / execution:** `s=Stack();s.push(1);s.push(2);print(s.pop(),s.pop())`

**Expected output:** `2 1`

**Dry-run trace:** head 1 -> head 2->1 -> head 1 -> None

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Linked Stack.
Algorithm: Use list head as stack top.
Logic: Push/pop in O(1).
"""

class Node:
    def __init__(self,x,n=None):self.x=x;self.next=n
class Stack:
    def __init__(self):self.head=None
    def push(self,x):self.head=Node(x,self.head)
    def pop(self):
        if self.head is None:raise IndexError('underflow')
        x=self.head.x;self.head=self.head.next;return x


s=Stack();s.push(1);s.push(2);print(s.pop(),s.pop())
```

## 03. Linked Queue

**Source classification:** Linked queue

**Algorithm:** Keep front and rear pointers to linked nodes.

**Implementation logic:** Enqueue and dequeue in O(1).

**Example input / execution:** `q=Queue();q.enqueue(8);q.enqueue(9);print(q.dequeue(),q.dequeue())`

**Expected output:** `8 9`

**Dry-run trace:** front/rear 8 -> front8 rear9 -> front9 rear9 -> both None

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Linked Queue.
Algorithm: Keep front and rear pointers to linked nodes.
Logic: Enqueue and dequeue in O(1).
"""

class Node:
    def __init__(self,x):self.x=x;self.next=None
class Queue:
    def __init__(self):self.front=self.rear=None
    def enqueue(self,x):
        n=Node(x)
        if self.rear:self.rear.next=n
        else:self.front=n
        self.rear=n
    def dequeue(self):
        if self.front is None:raise IndexError('underflow')
        x=self.front.x;self.front=self.front.next
        if self.front is None:self.rear=None
        return x


q=Queue();q.enqueue(8);q.enqueue(9);print(q.dequeue(),q.dequeue())
```

## 04. Reverse Linked List

**Source classification:** Integrated challenge

**Algorithm:** Reverse next pointers in one pass.

**Implementation logic:** Use prev, curr, nxt to preserve remaining nodes.

**Example input / execution:** `h=Node(1,Node(2,Node(3)));print(values(reverse(h)))`

**Expected output:** `[3, 2, 1]`

**Dry-run trace:** prev:None->1->2->3; curr:1->2->3->None

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Reverse Linked List.
Algorithm: Reverse next pointers in one pass.
Logic: Use prev, curr, nxt to preserve remaining nodes.
"""

class Node:
    def __init__(self,x,n=None):self.x=x;self.next=n
def reverse(head):
    prev=None;curr=head
    while curr:
        nxt=curr.next;curr.next=prev;prev=curr;curr=nxt
    return prev
def values(h):
    a=[]
    while h:a.append(h.x);h=h.next
    return a


h=Node(1,Node(2,Node(3)));print(values(reverse(h)))
```

## 05. Middle Of Linked List

**Source classification:** Section 21

**Algorithm:** Use slow and fast references.

**Implementation logic:** Fast advances twice as quickly; second middle is returned for even lengths.

**Example input / execution:** `print(middle(Node(1,Node(2,Node(3,Node(4,Node(5)))))))`

**Expected output:** `3`

**Dry-run trace:** slow 1->2->3; fast 1->3->5

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Middle Of Linked List.
Algorithm: Use slow and fast references.
Logic: Fast advances twice as quickly; second middle is returned for even lengths.
"""

class Node:
    def __init__(self,x,n=None):self.x=x;self.next=n
def middle(head):
    slow=fast=head
    while fast and fast.next:slow=slow.next;fast=fast.next.next
    return slow.x if slow else None


print(middle(Node(1,Node(2,Node(3,Node(4,Node(5)))))))
```

## 06. Cycle Detection

**Source classification:** Extended practice

**Algorithm:** Floyd tortoise/hare test.

**Implementation logic:** If fast catches slow, the list contains a cycle.

**Example input / execution:** `a,b,c=Node(1),Node(2),Node(3);a.next=b;b.next=c;c.next=b;print(has_cycle(a));c.next=None;print(has_cycle(a))`

**Expected output:** `True / False`

**Dry-run trace:** cycle 2->3->2 causes slow and fast to meet

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Cycle Detection.
Algorithm: Floyd tortoise/hare test.
Logic: If fast catches slow, the list contains a cycle.
"""

class Node:
    def __init__(self,x):self.x=x;self.next=None
def has_cycle(head):
    slow=fast=head
    while fast and fast.next:
        slow=slow.next;fast=fast.next.next
        if slow is fast:return True
    return False


a,b,c=Node(1),Node(2),Node(3);a.next=b;b.next=c;c.next=b;print(has_cycle(a));c.next=None;print(has_cycle(a))
```

## 07. Circular Singly Linked List

**Source classification:** Circular singly list

**Algorithm:** Keep tail whose next points at head.

**Implementation logic:** O(1) end insert and front delete.

**Example input / execution:** `c=CircularList();[c.append(x) for x in (1,2,3)];print(c.pop_front(),c.values())`

**Expected output:** `1 [2, 3]`

**Dry-run trace:** tail3 points head1, then after delete points head2

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Circular Singly Linked List.
Algorithm: Keep tail whose next points at head.
Logic: O(1) end insert and front delete.
"""

class Node:
    def __init__(self,x):self.x=x;self.next=None
class CircularList:
    def __init__(self):self.tail=None
    def append(self,x):
        n=Node(x)
        if self.tail:n.next=self.tail.next;self.tail.next=n
        else:n.next=n
        self.tail=n
    def pop_front(self):
        if not self.tail:raise IndexError('empty')
        h=self.tail.next;x=h.x
        if h is self.tail:self.tail=None
        else:self.tail.next=h.next
        return x
    def values(self):
        if not self.tail:return []
        ans=[];p=self.tail.next
        while True:
            ans.append(p.x);p=p.next
            if p is self.tail.next:break
        return ans


c=CircularList();[c.append(x) for x in (1,2,3)];print(c.pop_front(),c.values())
```

## 08. Doubly Linked List

**Source classification:** Doubly linked list

**Algorithm:** Maintain prev/next links and head/tail.

**Implementation logic:** O(1) operations at both ends.

**Example input / execution:** `d=DoublyList();[d.append(x) for x in (5,6,7)];print(d.pop_back(),d.values())`

**Expected output:** `7 [5, 6]`

**Dry-run trace:** 5<->6<->7 -> 5<->6

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Doubly Linked List.
Algorithm: Maintain prev/next links and head/tail.
Logic: O(1) operations at both ends.
"""

class Node:
    def __init__(self,x):self.x=x;self.prev=self.next=None
class DoublyList:
    def __init__(self):self.head=self.tail=None
    def append(self,x):
        n=Node(x);n.prev=self.tail
        if self.tail:self.tail.next=n
        else:self.head=n
        self.tail=n
    def pop_back(self):
        if not self.tail:raise IndexError('empty')
        x=self.tail.x;self.tail=self.tail.prev
        if self.tail:self.tail.next=None
        else:self.head=None
        return x
    def values(self):
        p=self.head;r=[]
        while p:r.append(p.x);p=p.next
        return r


d=DoublyList();[d.append(x) for x in (5,6,7)];print(d.pop_back(),d.values())
```

## 09. Circular Doubly Linked List

**Source classification:** Circular doubly linked list

**Algorithm:** Keep head and tail with circular bidirectional links.

**Implementation logic:** Insert and remove at front with constant pointer changes.

**Example input / execution:** `d=CircularDoubly();[d.append(x) for x in (1,2,3)];print(d.pop_front(),d.values())`

**Expected output:** `1 [2, 3]`

**Dry-run trace:** head1, tail3; reconnect 3<->2 after removing 1

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Circular Doubly Linked List.
Algorithm: Keep head and tail with circular bidirectional links.
Logic: Insert and remove at front with constant pointer changes.
"""

class Node:
    def __init__(self,x):self.x=x;self.next=self.prev=self
class CircularDoubly:
    def __init__(self):self.head=None
    def append(self,x):
        n=Node(x)
        if not self.head:self.head=n;return
        tail=self.head.prev;n.prev=tail;n.next=self.head;tail.next=n;self.head.prev=n
    def pop_front(self):
        if not self.head:raise IndexError('empty')
        h=self.head;x=h.x
        if h.next is h:self.head=None
        else:
            tail=h.prev;self.head=h.next;tail.next=self.head;self.head.prev=tail
        return x
    def values(self):
        if not self.head:return []
        p=self.head;a=[]
        while True:
            a.append(p.x);p=p.next
            if p is self.head:return a


d=CircularDoubly();[d.append(x) for x in (1,2,3)];print(d.pop_front(),d.values())
```

## 10. Header Sentinel List

**Source classification:** Header nodes

**Algorithm:** Maintain a dummy head to simplify edge cases.

**Implementation logic:** Insert/delete uniformly via predecessor.

**Example input / execution:** `h=HeaderList();h.prepend(2);h.prepend(1);print(h.delete(2),h.values())`

**Expected output:** `True [1]`

**Dry-run trace:** header -> 1 -> 2; bypass 2

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Header Sentinel List.
Algorithm: Maintain a dummy head to simplify edge cases.
Logic: Insert/delete uniformly via predecessor.
"""

class Node:
    def __init__(self,x=None,n=None):self.x=x;self.next=n
class HeaderList:
    def __init__(self):self.header=Node()
    def prepend(self,x):self.header.next=Node(x,self.header.next)
    def delete(self,x):
        p=self.header
        while p.next and p.next.x!=x:p=p.next
        if p.next:p.next=p.next.next;return True
        return False
    def values(self):
        a=[];p=self.header.next
        while p:a.append(p.x);p=p.next
        return a


h=HeaderList();h.prepend(2);h.prepend(1);print(h.delete(2),h.values())
```

## 11. Delete Given Node

**Source classification:** Integrated challenge 4

**Algorithm:** Copy successor payload and bypass successor.

**Implementation logic:** Only works for non-tail singly linked nodes.

**Example input / execution:** `h=Node(1,Node(2,Node(3)));delete_given(h.next);print([h.x,h.next.x])`

**Expected output:** `[1, 3]`

**Dry-run trace:** copy next value 3 into node 2, bypass old node 3

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Delete Given Node.
Algorithm: Copy successor payload and bypass successor.
Logic: Only works for non-tail singly linked nodes.
"""

class Node:
    def __init__(self,x,n=None):self.x=x;self.next=n
def delete_given(node):
    if node is None or node.next is None:raise ValueError('cannot delete tail')
    node.x=node.next.x;node.next=node.next.next


h=Node(1,Node(2,Node(3)));delete_given(h.next);print([h.x,h.next.x])
```
