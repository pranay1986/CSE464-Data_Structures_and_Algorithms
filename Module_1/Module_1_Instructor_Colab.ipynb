## 01. Array Operations

**Source classification:** Core implementation

**Algorithm:** Traverse, insert at an index by right-shifting, delete by left-shifting.

**Implementation logic:** Explicit shifts illustrate classical array costs.

**Example input / execution:** `a=[10,20,30,None,None];n=3;n=insert(a,n,15,1);print(a[:n]);n,x=delete(a,n,2);print(x,a[:n])`

**Expected output:** `[10, 15, 20, 30] / 20 [10, 15, 30]`

**Dry-run trace:** Follow the variables and intermediate states printed by the example.

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Array Operations.
Algorithm: Traverse, insert at an index by right-shifting, delete by left-shifting.
Logic: Explicit shifts illustrate classical array costs.
"""

def insert(a,n,x,i):
    if not 0<=i<=n<len(a): raise IndexError('invalid position or full array')
    for k in range(n,i,-1): a[k]=a[k-1]
    a[i]=x
    return n+1

def delete(a,n,i):
    if not 0<=i<n: raise IndexError(i)
    x=a[i]
    for k in range(i,n-1): a[k]=a[k+1]
    a[n-1]=None
    return n-1,x


a=[10,20,30,None,None];n=3;n=insert(a,n,15,1);print(a[:n]);n,x=delete(a,n,2);print(x,a[:n])
```

## 02. Linear Search

**Source classification:** Six-mark question 13

**Algorithm:** Inspect elements from left to right until the first match.

**Implementation logic:** Return index or -1, counting probes.

**Example input / execution:** `print(linear_search([8,3,19,14,6,22],14))`

**Expected output:** `3`

**Dry-run trace:** i=0:8; i=1:3; i=2:19; i=3:14 found

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Linear Search.
Algorithm: Inspect elements from left to right until the first match.
Logic: Return index or -1, counting probes.
"""

def linear_search(a,key):
    for i,v in enumerate(a):
        if v==key:return i
    return -1


print(linear_search([8,3,19,14,6,22],14))
```

## 03. Recursive Linear Search

**Source classification:** Six-mark question 14

**Algorithm:** Recursively inspect one item per call.

**Implementation logic:** Advance the index; stop when found or exhausted.

**Example input / execution:** `print(search([8,3,19,14,6,22],14))`

**Expected output:** `3`

**Dry-run trace:** indices 0, 1, 2, 3; recursion depth 4

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Recursive Linear Search.
Algorithm: Recursively inspect one item per call.
Logic: Advance the index; stop when found or exhausted.
"""

def search(a,key,i=0):
    if i==len(a):return -1
    if a[i]==key:return i
    return search(a,key,i+1)


print(search([8,3,19,14,6,22],14))
```

## 04. Binary Search

**Source classification:** Six-mark question 15

**Algorithm:** Halve a sorted search interval until matching or empty.

**Implementation logic:** Maintain inclusive low/high boundaries.

**Example input / execution:** `print(binary_search([5,9,13,20,27,31,38,42,50,61],42))`

**Expected output:** `7`

**Dry-run trace:** [0,9] mid=4 (27), [5,9] mid=7 (42)

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Binary Search.
Algorithm: Halve a sorted search interval until matching or empty.
Logic: Maintain inclusive low/high boundaries.
"""

def binary_search(a,key):
    lo,hi=0,len(a)-1
    while lo<=hi:
        mid=(lo+hi)//2
        if a[mid]==key:return mid
        if a[mid]<key:lo=mid+1
        else:hi=mid-1
    return -1


print(binary_search([5,9,13,20,27,31,38,42,50,61],42))
```

## 05. Recursive Binary Search

**Source classification:** Six-mark question 17

**Algorithm:** Recursively halve sorted range.

**Implementation logic:** Avoid slicing so only logarithmic call frames are used.

**Example input / execution:** `print(search([2,5,8,11,19,24],19))`

**Expected output:** `4`

**Dry-run trace:** mid 2 -> mid 4 found

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Recursive Binary Search.
Algorithm: Recursively halve sorted range.
Logic: Avoid slicing so only logarithmic call frames are used.
"""

def search(a,x,lo=0,hi=None):
    if hi is None:hi=len(a)-1
    if lo>hi:return -1
    mid=(lo+hi)//2
    if a[mid]==x:return mid
    if a[mid]<x:return search(a,x,mid+1,hi)
    return search(a,x,lo,mid-1)


print(search([2,5,8,11,19,24],19))
```

## 06. First Occurrence

**Source classification:** Challenge 1

**Algorithm:** Modified binary search continues left after a match.

**Implementation logic:** Record the candidate index.

**Example input / execution:** `print(first([2,4,4,4,7,9,12],4))`

**Expected output:** `1`

**Dry-run trace:** mid=3 found -> left; mid=1 found -> left; mid=0 smaller; return 1

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | First Occurrence.
Algorithm: Modified binary search continues left after a match.
Logic: Record the candidate index.
"""

def first(a,x):
    lo,hi,ans=0,len(a)-1,-1
    while lo<=hi:
        mid=(lo+hi)//2
        if a[mid]>=x:
            if a[mid]==x:ans=mid
            hi=mid-1
        else:lo=mid+1
    return ans


print(first([2,4,4,4,7,9,12],4))
```

## 07. Insertion Position

**Source classification:** Challenge 2

**Algorithm:** Lower-bound binary search.

**Implementation logic:** Return the first index whose value is not smaller than target.

**Example input / execution:** `print([insertion_index([3,8,12,20,31],x) for x in (12,15,40)])`

**Expected output:** `[2, 3, 5]`

**Dry-run trace:** 15: midpoint 2=12 -> lo=3; midpoint 4=31 -> hi=4; midpoint 3=20 -> hi=3

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Insertion Position.
Algorithm: Lower-bound binary search.
Logic: Return the first index whose value is not smaller than target.
"""

def insertion_index(a,x):
    lo,hi=0,len(a)
    while lo<hi:
        mid=(lo+hi)//2
        if a[mid]<x:lo=mid+1
        else:hi=mid
    return lo


print([insertion_index([3,8,12,20,31],x) for x in (12,15,40)])
```

## 08. Count Occurrences

**Source classification:** Challenge 3

**Algorithm:** Compute lower/upper bounds to count duplicates.

**Implementation logic:** Two logarithmic binary searches.

**Example input / execution:** `print(count([1,2,2,2,2,5,7,7,9],2))`

**Expected output:** `4`

**Dry-run trace:** lower_bound=1, upper_bound=5, count=4

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Count Occurrences.
Algorithm: Compute lower/upper bounds to count duplicates.
Logic: Two logarithmic binary searches.
"""

def bound(a,x,strict):
    lo,hi=0,len(a)
    while lo<hi:
        mid=(lo+hi)//2
        if a[mid]<x or (strict and a[mid]==x):lo=mid+1
        else:hi=mid
    return lo

def count(a,x):return bound(a,x,True)-bound(a,x,False)


print(count([1,2,2,2,2,5,7,7,9],2))
```

## 09. Loop Growth Counts

**Source classification:** Six/ten-mark loop analysis

**Algorithm:** Instrument representative linear, logarithmic, triangular loops.

**Implementation logic:** Count operations without relying on wall-clock time.

**Example input / execution:** `print(counts(8))`

**Expected output:** `(8, 4, 36)`

**Dry-run trace:** linear 8; i:1,2,4,8; triangular 1+...+8=36

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Loop Growth Counts.
Algorithm: Instrument representative linear, logarithmic, triangular loops.
Logic: Count operations without relying on wall-clock time.
"""

def counts(n):
    linear=sum(1 for _ in range(n))
    logarithmic=0;i=1
    while i<=n:logarithmic+=1;i*=2
    triangular=sum(1 for i in range(1,n+1) for _ in range(i))
    return linear,logarithmic,triangular


print(counts(8))
```

## 10. Time Space Tradeoff

**Source classification:** Section 10

**Algorithm:** Compare direct repeated computation with a precomputed lookup table.

**Implementation logic:** Cache squares in O(n) extra memory.

**Example input / execution:** `q=[3,5,3,5];print(uncached(q)==cached(q),cached(q))`

**Expected output:** `True [9, 25, 9, 25]`

**Dry-run trace:** table {3:9,5:25}; queries read 9,25,9,25

**Complexity exercise:** Derive the worst-case time and auxiliary space from the implementation; identify edge cases.


```python
"""CSE464 | Time Space Tradeoff.
Algorithm: Compare direct repeated computation with a precomputed lookup table.
Logic: Cache squares in O(n) extra memory.
"""

def uncached(queries):return [x*x for x in queries]
def cached(queries):
    table={x:x*x for x in set(queries)}
    return [table[x] for x in queries]


q=[3,5,3,5];print(uncached(q)==cached(q),cached(q))
```
